## 2. Trabalhando com listas

Uma lista pode guardar todas as notas de uma turma.


```python
notas = [7.0, 8.5, 6.0, 9.0]
print("Primeira nota:", notas[0])
print("Quantidade:", len(notas))
print("Soma:", sum(notas))
```

## 3. Calculando uma média

Combine `sum()` e `len()` para calcular a média de uma lista não vazia.


```python
notas = [7, 8, 6, 9]
media = sum(notas) / len(notas)
print(f"Média: {media:.2f}")
```

## 4. Crie suas próprias ferramentas

Funções permitem reaproveitar uma lógica. `return` devolve um resultado para quem chamou a função.


```python
def calcular_media(notas):
    return sum(notas) / len(notas)

resultado = calcular_media([8, 7, 9])
print(resultado)
```
